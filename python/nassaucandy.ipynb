```python
# Create reusable functions
import re

def standardize_column_names(df):

    df = df.copy()

    new_columns = []

    for column in df.columns:

        column = str(column).strip()

        # Replace special characters and spaces with underscores
        column = re.sub(
            r"[^A-Za-z0-9]+",
            "_",
            column
        )

        # Remove underscores at the beginning and end
        column = column.strip("_")

        # Convert to lowercase
        column = column.lower()

        new_columns.append(column)

    df.columns = new_columns

    return df
```

```python
for table_name in clean_data:

    clean_data[table_name] = standardize_column_names(
        clean_data[table_name]
    )

    print(f"\n{table_name}")
    print(clean_data[table_name].columns.tolist())
```

```python
# create a text-cleaning functions
def clean_text_columns(df):

    df = df.copy()

    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for column in text_columns:

        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
        )

        # Convert empty strings to missing values
        df[column] = df[column].replace("", pd.NA)

    return df
```

```python

#  Business KPI Calculations

def safe_divide(numerator, denominator):
    """
    Safely divide two values.
    Returns 0 if denominator is zero.
    """
    if denominator == 0:
        return 0
    return numerator / denominator


def calculate_business_kpis(data, order_data):
    """
    Calculate important business KPIs using:
    1. Transaction-level analytical data
    2. Order-level aggregated data
    """

    total_sales = data["sales"].sum()
    total_orders = data["order_id"].nunique()
    total_units = data["units"].sum()
    total_gross_profit = data["gross_profit"].sum()
    total_cost = data["cost"].sum()

    unique_customers = data["customerkey"].nunique()
    unique_products = data["productkey"].nunique()
    unique_factories = data["factorykey"].nunique()
    unique_cities = data["city"].nunique()
    unique_routes = data["route_id"].nunique()

    average_order_value = safe_divide(
        total_sales,
        total_orders
    )

    gross_profit_margin = safe_divide(
        total_gross_profit,
        total_sales
    ) * 100

    cost_to_sales_ratio = safe_divide(
        total_cost,
        total_sales
    ) * 100

    average_units_per_order = safe_divide(
        total_units,
        total_orders
    )

    sales_per_customer = safe_divide(
        total_sales,
        unique_customers
    )

    gross_profit_per_order = safe_divide(
        total_gross_profit,
        total_orders
    )

    average_transaction_value = safe_divide(
        total_sales,
        len(data)
    )

    return {
        "Total Transaction Rows": len(data),
        "Total Orders": total_orders,
        "Total Customers": unique_customers,
        "Total Products": unique_products,
        "Total Factories": unique_factories,
        "Total Cities": unique_cities,
        "Total Routes": unique_routes,
        "Total Sales": total_sales,
        "Total Units": total_units,
        "Total Gross Profit": total_gross_profit,
        "Total Cost": total_cost,
        "Average Order Value": average_order_value,
        "Average Transaction Value": average_transaction_value,
        "Gross Profit Margin (%)": gross_profit_margin,
        "Cost to Sales Ratio (%)": cost_to_sales_ratio,
        "Average Units per Order": average_units_per_order,
        "Sales per Customer": sales_per_customer,
        "Gross Profit per Order": gross_profit_per_order
    }
```

```python

# Reusable KPI card function

def display_kpi_card(title, value, format_type="number"):

    if format_type == "currency":
        display_value = f"{value:,.2f}"

    elif format_type == "percentage":
        display_value = f"{value:.2f}%"

    elif format_type == "integer":
        display_value = f"{value:,.0f}"

    else:
        display_value = f"{value:,.2f}"

    st.metric(
        label=title,
        value=display_value
    )
```

```python
display_kpi_card("Total Sales", 141783.63)
display_kpi_card("Total Orders", 8549, "integer")
display_kpi_card("Gross Profit Margin", 65.91, "percentage")
```

```python

# Reusable Plotly bar chart function

def create_bar_chart(
    data,
    x_column,
    y_column,
    chart_title,
    x_label,
    y_label
):

    fig = px.bar(
        data,
        x=x_column,
        y=y_column,
        title=chart_title,
        labels={
            x_column: x_label,
            y_column: y_label
        }
    )

    fig.update_layout(
        template="plotly_white",
        height=450
    )

    return fig
```
